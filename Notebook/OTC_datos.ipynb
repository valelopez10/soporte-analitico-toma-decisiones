# Procesamiento de datos de colaboradores OTC

Lee `Catalogo_Colaboradores.xlsx`, elimina duplicados y registros incompletos, calcula las columnas de salario y guarda `archivo_actualizado.xlsx`.

```python
from pathlib import Path

import pandas as pd

CARPETA = Path.cwd()
if not (CARPETA / "Catalogo_Colaboradores.xlsx").exists() and (
    CARPETA.parent / "Catalogo_Colaboradores.xlsx"
).exists():
    CARPETA = CARPETA.parent

ARCHIVO_ENTRADA = CARPETA / "Catalogo_Colaboradores.xlsx"
ARCHIVO_SALIDA = CARPETA / "archivo_actualizado.xlsx"
COLUMNAS_REQUERIDAS = (
    "Nombre Completo",
    "Sueldo Neto IMSS",
    "Sueldo Complementos",
    "Sueldo Neto Total",
)

if not ARCHIVO_ENTRADA.exists():
    raise FileNotFoundError(f"No se encontró el archivo de entrada: {ARCHIVO_ENTRADA}")

df = pd.read_excel(ARCHIVO_ENTRADA)
faltantes = [columna for columna in COLUMNAS_REQUERIDAS if columna not in df.columns]
if faltantes:
    raise ValueError(f"Faltan columnas requeridas en el Excel: {faltantes}")

df = df.drop_duplicates().dropna(subset=COLUMNAS_REQUERIDAS).copy()
df["Nombre Completo"] = df["Nombre Completo"].astype("string").str.strip().str.upper()
df = df[df["Nombre Completo"].ne("")].copy()

df["SalarioDiario"] = df["Sueldo Neto Total"] / 192
df["CostoInterno"] = df["Sueldo Neto IMSS"] + df["Sueldo Complementos"]

df.to_excel(ARCHIVO_SALIDA, index=False)
print(f"Archivo actualizado: {ARCHIVO_SALIDA}")
print(f"Registros procesados: {len(df)}")
```

```python
estadisticas = df[[
    "SalarioDiario",
    "Sueldo Neto IMSS",
    "Sueldo Complementos",
    "CostoInterno",
]].mean()
estadisticas
```
